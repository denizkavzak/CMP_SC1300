```python
def getAverage(values):
    total = 0
    for i in range(len(values)):
        total += values[i]
    if len(values) == 0:
        return 0
    return total / len(values)

l = [100, 50, 20, 200, 75, 10, 5]
print(getAverage(l))
print(getAverage([]))
```
