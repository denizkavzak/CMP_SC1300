```python
j = 2
result = 0
number = int(input("Please enter an integer (2 or greater):"))
while j < number :  
   if number % j  == 0 :
      result = 1
      break
   j = j + 1
if result == 1 :
   print("Number:", number, "is Not Prime.")
else :
   print("Number:", number, "is Prime.")
```
