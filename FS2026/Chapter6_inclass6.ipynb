```python
def main():
  table1 = [ [10,20,30,40,50], [100,200,300,400,500], [15,30,45,60,75]   ]

  print(len(table1))
  print(table1[0])
  print(len(table1[0]))

  result = sumTableAll(table1)
  print(result)


def sumTableAll(values):
  row = len(values)
  col = len(values[0])
  total = 0
  for i in range(row):
    for j in range(col):
      total += values[i][j]
  return total

main()
```

```python
def main():
  table1 = [ [75,80,90], [100,75,87], [55,75,70], [90, 50, 70], [65, 100, 70] ]

  averages = getAverageGradesAllStudents(table1)
  print(averages)

## find the average grade of each student and return it as a list
def getAverageGradesAllStudents(grades):
  row = len(grades)
  col = len(grades[0])

  averages = []

  for i in range(row): # this represents student number i at each loop
    total = 0 # for each student, initialize to 0
    for j in range(col): # this represents grades of student i
      #sum(grades[i])
      total += grades[i][j]

    avg = total / col
    averages.append(avg)

  return averages


main()

```
