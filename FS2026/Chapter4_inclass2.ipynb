```python
grade = int(input("Enter a grade (0-100), -1 to stop:"))
countPasses = 0
countFails = 0

while grade != -1:
    if grade < 60:
        countFails += 1
    else:
        countPasses += 1
    grade = int(input("Enter a grade (0-100), -1 to stop:"))

print("Fails:", countFails)
print("Fails:", countPasses)
```

```python
numOfStudents = int(input("Enter number of students:"))
numOfGrades = int(input("Enter number of grades:"))
studentCount = 1
gradeCount = 1
#total = 0
countPasses = 0
countFails = 0

while studentCount <= numOfStudents: # this will run for each student
    while gradeCount <= numOfGrades: # this will run 3 grades for each student
        grade = int(input("Enter grade (0-100):"))
        if grade < 60:
            countFails += 1
        else:
            countPasses += 1
        gradeCount += 1
    gradeCount = 1
    studentCount += 1
        
print("Fails:", countFails)
print("Passes:", countPasses)
```
