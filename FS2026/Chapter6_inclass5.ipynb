```python
def main():
    numbers = []
    num = input("Enter a number, Q to stop")
    while num != "Q":
        numbers.append(int(num))
        num = input("Enter a number, Q to stop")

    print(getAverage(numbers))

def getAverage(numbers):
    if len(numbers) == 0:
        print("List is empty")
        return -1
    total = 0
    for number in numbers:
        total += number
    return total/len(numbers)

main()
```

```python
##
# This function will create a new list
# with elements from 1 to n (n included)
# multiplied by 2
##
def createListDoubleN(n):
    numbers = []
    for i in range(1,n+1):
        numbers.append(i*2)
    return numbers
        
result = createListDoubleN(10)
print(result)
```

```python
## this function will take a list 
# calculate the average, min and max
# and return all 3 values as a tuple
def getStatsOfList(numbers):
    total = 0
    max = min = numbers[0]
    for number in numbers:
        total += number
        if number < min:
            min = number
        if number > max:
            max = number
    
    average = total/len(numbers)
    
    return (average, min, max)

values = [100,40,25,40,150]

result = getStatsOfList(values) # result is a tuple
print(result)

print(result[0])

averageResult, minResult, maxResult = getStatsOfList(values)
print(averageResult)
print(minResult)
print(maxResult)

averageResult, _ , _  = getStatsOfList(values)
print(averageResult)
```

```python
def main():
    numbers = [1,2,3,4,5]
    print(numbers)
    doubleListElement(numbers)
    print(numbers)

    numbersT = (1,2,3,4,5)
    doubleTupleElement(numbersT)
    print(numbersT)

def doubleListElement(values):
    for i in range(len(values)):
        values[i] = values[i] * 2

def doubleTupleElement(valuesT):
    for i in range(len(valuesT)):
        valuesT[i] = valuesT[i] * 2

main()
```
