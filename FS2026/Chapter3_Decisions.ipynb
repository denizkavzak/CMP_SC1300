```python
from math import sqrt

EPSILON = 1E-14

r = sqrt(2.0)

if abs(r*r - 2.0) < EPSILON:
  print("sqrt(2.0) squared is approximately 2.0")
else:
  print("sqrt(2.0) squared is not approximately 2.0 but", r*r)
```

```python
## A movie theater example
# Child, regular, senior ticket

age = int(input("Enter age: "))

if age < 13:
  print("You qualify for a child ticket")
  ticketPrice = 8
elif age < 65:
  print("Regular ticket")
  ticketPrice = 12
else:
  print("Senior ticket")
  ticketPrice = 9

is_student = True
if is_student:
  print("Student discount")
  ticketPrice = ticketPrice * 0.8

# examples with strings
day = "Tuesday"
if day == "Tuesday":
  print("Discount day")

membership = input("Membership type: ")

if membership.lower() == "premium":
    print("Premium discount applied")
    ticketPrice = ticketPrice * 0.7

# nested if blocks
if age >= 18:
  if is_student:
    print("Adult student discount")
  else:
    print("Regular adult ticket")

# boolean operators
if age < 13 or age >= 65:
  print("Discount ticket")

if age >= 13 and age < 65:
  print("Regular ticket")

# some input validation
if age < 0 or age > 120:
    print("Invalid age")
else:
    print("Valid age")
```
