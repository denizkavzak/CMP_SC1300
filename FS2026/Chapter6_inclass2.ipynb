```python
grade = int(input("Enter grades, -1 to stop"))
grades = []
print(len(grades))
while grade != -1:
    grades.append(grade)
    grade = int(input("Enter grades, -1 to stop"))

print(sum(grades)) # sum works with empty list

print(grades)
if len(grades) != 0:
    print("average is:", sum(grades)/len(grades))
    print("max is:", max(grades)) # min and max gives error with empty list
    print("min is:", min(grades))
else:
    print("No grades were entered")
```

```python
def main():
    l = [1,2,3,4,5,0]
    l = []
    print(getMin(l))

    if len(l) != 0:
        print("min is:", min(l))
##
# Gets the minimum value of a list of positive numbers
##
def getMin(list):
    if len(list) == 0:
        return -1 # if the list is empty, return -1
    min = list[0]
    for i in range(1, len(list)):
        if list[i] < min:
            min = list[i]
    return min

main()
```

```python
grade = int(input("Enter grades, -1 to stop"))
grades = []
while grade != -1:
    grades.append(grade)
    grade = int(input("Enter grades, -1 to stop"))

countPasses = 0
countFails = 0

for grade in grades:
    if grade >= 60:
        countPasses = countPasses + 1
    else:
        countFails = countFails + 1

print(countPasses)
print(countFails)
```
