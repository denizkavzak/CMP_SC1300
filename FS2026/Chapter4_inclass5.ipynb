```python
numOfStudents = int(input("Enter number of students:"))

for i in range(numOfStudents):
    total = 0
    grade = int(input("Enter a grade 0-100, -1 to stop"))
    while grade != -1:
        total = total + grade
        grade = int(input("Enter a grade 0-100, -1 to stop"))
    print("Total grade for student no", i, "is", total)
```

```python
moreStudentCheck = int(input("Enter grades for a student, 'N' to stop:"))

while moreStudentCheck != 'N':
    total = 0
    grade = int(input("Enter a grade 0-100, -1 to stop"))
    while grade != -1:
        total = total + grade
        grade = int(input("Enter a grade 0-100, -1 to stop"))
    print("Total grade for student no", i, "is", total)
    moreStudentCheck = input("Enter grades for a student, 'N' to stop:")
```

```python
password = "abcd1234"
digitCount = 0

for ch in password:
  if ch.isdigit():
    digitCount = digitCount + 1

print(digitCount)
```

```python
password = "abcd"
found = False

for ch in password:
  if ch.isdigit():
    found = True
    break

if found:
  print("Digit found")
else:
  print("No digits")
```

```python
password = "abcd1234"
found = False
i = 0

while i < len(password) and not found:
  if password[i].isdigit():
    found = True
  i = i + 1

if found:
  print("Digit found")
else:
  print("No digits")
```

```python
word = "Hello BYE byE goOd Night"
vowels = 0

for char in word:
  if char in "aeiouAEIOU":
    vowels = vowels + 1

print(vowels)
```

```python
word = "Hello BYE byE goOd Night"
vowels = 0

for char in word:
  if char.lower() in "aeiou":
    vowels = vowels + 1

print(vowels)
```
