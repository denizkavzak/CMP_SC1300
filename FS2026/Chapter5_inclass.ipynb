```python
##
# Calculates the volume of a cube
# @ param sideLength is the side length of cube
# @ return volume of cube
def getVolume(sideLength):
  return sideLength ** 3

def printVolume(sideLength):
  print("Volume of side length 5 is", getVolume(sideLength))

print(getVolume(5))
print(getVolume(10))

print("Volume of side length 5 is", getVolume(5))
print("Volume of side length 10 is", getVolume(10))

printVolume(5)
printVolume(10)
```

```python
##
# Search for a char in a string
# return True if exists, return false
# if it doesn't exist
# @ param string1 original string and
# searchChar is the searched character
# @ return True if found, False if not
def searchString(string1, searchChar):
  for ch in string1:
    if ch == searchChar:
      #print("Found")
      return True
  #print("not found")
  return False

string1 = "Hello bye bye"
searchChar = "A"

found = searchString(string1, searchChar)
if found:
  print("found")
else:
  print("not found")

found = searchString(string1, 'y') # expected output is True
if found:
  print("Test passed for y")

found = searchString(string1, 'E') # expected output is False
if not found:
  print("Test passed for E")

found = searchString(string1, 'b') # expected output is True
if found:
  print("Test passed for b")
```

```python
numberOfElements = 10 # global variable

def main():
  printElements()
  printEvenNumbers()

def printElements():
  global numberOfElements
  for i in range(numberOfElements):
    print(i)

def printEvenNumbers():
  global numberOfElements
  for i in range(numberOfElements):
    if i % 2 == 0:
      print(i)

main()
```

```python
def fact(num):
  if num == 1:
    return 1
  else:
    return num * fact(num - 1)

print(fact(5))
```
