```python
num_of_grades = 3
counter = 1
total = 0 # to keep track of the sum

while counter <= num_of_grades:
    grade = int(input("Enter a grade: "))
    print("You entered: ", grade)
    total = total + grade # updating the sum
    counter += 1 # counter = counter + 1

average = total / num_of_grades
print("Average of the numbers entered: ", average )
print("done")
```

```python
num_of_grades = int(input("Enter the number of grades:"))
counter = 1
count_passes = 0
count_fails = 0

while counter <= num_of_grades:
    grade = int(input("Enter grade:"))
    counter = counter + 1
    if grade < 60:
        count_fails += 1
    else: 
        count_passes += 1
    
print("Number of fails: ", count_fails)
print("Number of passes: ", count_passes)
```

```python
num_of_grades = int(input("Enter the number of grades:"))
counter = 1
max_grade = 0
min_grade = 100

while counter <= num_of_grades:
    grade = int(input("Enter grade:"))
    if grade > max_grade:
        max_grade = grade
    if grade < min_grade:
        min_grade = grade
    counter = counter + 1

print("Max grade:", max_grade)
print("Min grade:", min_grade)
```
