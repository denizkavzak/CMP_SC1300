```python
# dictionary of student_id: list of grades of that student

gradebook = {}

student_id = 1
student_grades = [75, 80, 90]

gradebook[student_id] = student_grades

student_id = 2
student_grades = [55, 85, 80]

gradebook[student_id] = student_grades

student_id = 3
student_grades = [70, 60, 95]

gradebook[student_id] = student_grades


for item in gradebook.items():
    print(item[0], item[1])
```

```python
# dictionary of student_id: set of unique id of classes

gradebook = {}

student_id = 1
student_classes = {"A", "B", "C", "A"}

gradebook[student_id] = student_classes

student_id = 2
student_classes = {"A", "B", "C", "D", "B", "A"}

gradebook[student_id] = student_classes

student_id = 3
student_classes = {"A", "B", "C", "D", "E"}

gradebook[student_id] = student_classes

for item in gradebook.items():
    print(item[0], item[1])
```

```python
people = []

dict = {}
dict["name"] = "Deniz"
dict["last_name"] = "Kavzak Ufuktepe"
dict["age"] = 36
dict["nationality"] = "Turkish"

people.append(dict)

dict2 = {}
dict2["name"] = "Ekincan"
dict2["last_name"] = "Ufuktepe"
dict2["age"] = 37
dict2["nationality"] = "Turkish"

people.append(dict2)

dict3 = {}
dict3["name"] = "Britton"
dict3["last_name"] = "Ufuktepe"
dict3["age"] = 5
dict3["nationality"] = "American"

people.append(dict3)
    
for i in range(len(people)):
    person_details = people[i]
    print("Name is: " + person_details["name"] + ", Last name is: " + person_details["last_name"])

print()

for i in range(len(people)):
    person_details = people[i]
    for detail in person_details.items():
        print(detail[0], detail[1])
```
