```python
infile = open("input.txt", "r")

line = infile.readline() # this reads the \n at the end of each line

print(line)

infile.close()
```

```python
infile = open("input.txt", "r")

line = infile.readline()
value = float(line) # the \n at the end of each line is ignored after conversion

print(value + value)
print(line + value)

infile.close()
```

```python
infiletext = open("inputText.txt", "r")

linetext = infiletext.readline()

print(linetext)

infiletext.close()
```

```python
infiletext = open("inputText.txt", "r")

linetext = infiletext.readline()
print(linetext)

while linetext != "":
    linetext = infiletext.readline()
    print(linetext)

infiletext.close()
```
