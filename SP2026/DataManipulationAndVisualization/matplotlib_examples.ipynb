```python
## 1. Basic line plot
# x values, y values, and plt.plot()
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.show()
```

```python
## 2. Line plot with labels and title
# graphs should communicate clearly, not just display data
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)

plt.title("Simple Line Plot")
plt.xlabel("X values")
plt.ylabel("Y values")

plt.show()
```

```python
## 3. Customize line color, marker, and style
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 5, 7, 8, 11]

plt.plot(
    x,
    y,
    color="green",
    marker="o",
    linestyle="--",
    linewidth=2
)

plt.title("Customized Line Plot")
plt.xlabel("Day")
plt.ylabel("Sales")

plt.show()
```

```python
## Common style options for line plots:
linestyle="-"    # solid line
linestyle="--"   # dashed line
linestyle=":"    # dotted line

marker="o"       # circle
marker="s"       # square
marker="^"       # triangle
```

```python
## 4. Multiple lines on one chart
# label plus plt.legend() helps compare categories.
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]

store_a = [100, 120, 130, 125, 150]
store_b = [90, 110, 115, 140, 145]

plt.plot(days, store_a, marker="o", label="Store A")
plt.plot(days, store_b, marker="s", label="Store B")

plt.title("Sales Over Time")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python
## 5. Bar chart
# Bar charts are good for comparing categories.
import matplotlib.pyplot as plt

subjects = ["Math", "Science", "English", "History"]
scores = [88, 92, 79, 85]

plt.bar(subjects, scores)

plt.title("Average Test Scores")
plt.xlabel("Subject")
plt.ylabel("Score")

plt.show()
```

```python
## 6. Horizontal bar chart
# Horizontal bars are useful when category names are long.
import matplotlib.pyplot as plt

cities = ["Chicago", "Houston", "Phoenix", "Philadelphia"]
population = [2.7, 2.3, 1.6, 1.5]

plt.barh(cities, population)

plt.title("City Population")
plt.xlabel("Population in millions")
plt.ylabel("City")

plt.show()
```

```python
## 7. Scatter plot
# Scatter plots are great for relationships between two numeric variables.
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7]
exam_scores = [55, 60, 65, 70, 75, 82, 88]

plt.scatter(hours_studied, exam_scores)

plt.title("Study Time vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.show()
```

```python
## 8. Histogram
# Histograms show the distribution of one numeric variable.
import matplotlib.pyplot as plt

ages = [18, 19, 21, 21, 22, 23, 23, 24, 25, 27, 29, 30, 31, 35]

plt.hist(ages, bins=5)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
## 9. Pie chart
# Pie charts are best when showing parts of a whole, 
# but they can become hard to read with too many categories.
import matplotlib.pyplot as plt

labels = ["Rent", "Food", "Transport", "Savings"]
amounts = [1200, 500, 300, 700]

plt.pie(amounts, labels=labels, autopct="%1.1f%%")

plt.title("Monthly Budget")
plt.show()
```

```python
## 10. Figure size and grid lines
# figsize controls chart size, and grid(True) can improve readability.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [12000, 15000, 14000, 18000, 21000]

plt.figure(figsize=(8, 5))

plt.plot(months, revenue, marker="o")

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)

plt.show()
```

```python
## 11. Rotating axis labels
# plt.tight_layout() helps prevent labels from being cut off.
import matplotlib.pyplot as plt

products = ["Laptop Stand", "Wireless Keyboard", "Noise Cancelling Headphones", "USB-C Hub"]
sales = [120, 85, 60, 150]

plt.figure(figsize=(9, 5))

plt.bar(products, sales)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.show()
```

```python
## 12. Saving a plot as an image
# Useful for reports, slides, assignments, or dashboards.
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, marker="o")

plt.title("Saved Plot Example")
plt.xlabel("X")
plt.ylabel("Y")

plt.savefig("my_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## 13. Subplots: multiple charts in one figure
# fig, axes = plt.subplots() gives more control than using only plt.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [12000, 15000, 14000, 18000, 21000]
expenses = [8000, 9000, 9500, 10000, 12000]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(months, revenue, marker="o")
axes[0].set_title("Revenue")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Dollars")

axes[1].plot(months, expenses, marker="s")
axes[1].set_title("Expenses")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Dollars")

plt.tight_layout()
plt.show() 
```

```python
## 14. Subplots with a 2 by 2 layout
# This is a good bridge into dashboards.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

revenue = [12000, 15000, 14000, 18000, 21000]
expenses = [8000, 9000, 9500, 10000, 12000]
profit = [4000, 6000, 4500, 8000, 9000]
customers = [200, 240, 230, 280, 310]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(months, revenue, marker="o")
axes[0, 0].set_title("Revenue")

axes[0, 1].plot(months, expenses, marker="o")
axes[0, 1].set_title("Expenses")

axes[1, 0].plot(months, profit, marker="o")
axes[1, 0].set_title("Profit")

axes[1, 1].plot(months, customers, marker="o")
axes[1, 1].set_title("Customers")

for ax in axes.flat:
    ax.set_xlabel("Month")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
## 15. Annotating a chart
# annotations help tell a story with data.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [12000, 15000, 14000, 18000, 21000]

plt.figure(figsize=(8, 5))

plt.plot(months, revenue, marker="o")

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.annotate(
    "Highest revenue",
    xy=("May", 21000),
    xytext=("Mar", 22000),
    arrowprops={"arrowstyle": "->"}
)

plt.show()
```

```python
## 16. Error bars
# Useful when teaching uncertainty, measurement error, or confidence intervals.
import matplotlib.pyplot as plt

groups = ["A", "B", "C", "D"]
means = [72, 85, 78, 90]
errors = [4, 6, 3, 5]

plt.bar(groups, means, yerr=errors, capsize=5)

plt.title("Average Scores with Error Bars")
plt.xlabel("Group")
plt.ylabel("Average Score")

plt.show()
```

```python
## 17. Stacked bar chart
# stacked bars show both totals and category composition.
import matplotlib.pyplot as plt

quarters = ["Q1", "Q2", "Q3", "Q4"]

product_a = [100, 120, 140, 160]
product_b = [80, 90, 100, 130]
product_c = [50, 70, 90, 110]

plt.bar(quarters, product_a, label="Product A")
plt.bar(quarters, product_b, bottom=product_a, label="Product B")

bottom_for_c = [
    product_a[i] + product_b[i]
    for i in range(len(product_a))
]

plt.bar(quarters, product_c, bottom=bottom_for_c, label="Product C")

plt.title("Quarterly Sales by Product")
plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python
## 18. Box plot
# box plots summarize median, spread, and outliers.
import matplotlib.pyplot as plt

class_a = [78, 82, 85, 88, 90, 92, 95]
class_b = [70, 75, 80, 85, 90, 95, 100]
class_c = [60, 65, 70, 75, 80, 85, 90]

plt.boxplot([class_a, class_b, class_c], labels=["Class A", "Class B", "Class C"])

plt.title("Exam Score Distribution by Class")
plt.ylabel("Score")

plt.show()
```

```python
## 19. Heatmap using Matplotlib
# heatmaps are helpful for matrix-like data.
import matplotlib.pyplot as plt
import numpy as np

data = np.array([
    [80, 85, 90],
    [70, 75, 88],
    [92, 89, 95],
    [60, 72, 78]
])

students = ["Student 1", "Student 2", "Student 3", "Student 4"]
subjects = ["Math", "Science", "English"]

fig, ax = plt.subplots(figsize=(7, 5))

image = ax.imshow(data)

ax.set_xticks(np.arange(len(subjects)))
ax.set_yticks(np.arange(len(students)))

ax.set_xticklabels(subjects)
ax.set_yticklabels(students)

plt.colorbar(image, label="Score")

ax.set_title("Student Scores Heatmap")

plt.show()
```

```python
## 20. with simple data
# fill_between() visually emphasizes the range between two lines.
import matplotlib.pyplot as plt

dates = [
    "Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"
]

high_temp = [72, 75, 78, 80, 77, 74, 73]
low_temp = [55, 57, 60, 63, 61, 58, 56]

plt.figure(figsize=(9, 5))

plt.plot(dates, high_temp, marker="o", label="High")
plt.plot(dates, low_temp, marker="o", label="Low")

plt.fill_between(dates, low_temp, high_temp, alpha=0.2)

plt.title("Weekly Temperature Range")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.legend()
plt.grid(True)

plt.show()
```

```python
## 21. Sales CSV file
# Matplotlib is often used together with pandas.

# month,revenue,expenses
# Jan,12000,8000
# Feb,15000,9000
# Mar,14000,9500
# Apr,18000,10000
# May,21000,12000

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales.csv")

df["profit"] = df["revenue"] - df["expenses"]

plt.figure(figsize=(9, 5))

plt.plot(df["month"], df["revenue"], marker="o", label="Revenue")
plt.plot(df["month"], df["expenses"], marker="o", label="Expenses")
plt.plot(df["month"], df["profit"], marker="o", label="Profit")

plt.title("Business Performance")
plt.xlabel("Month")
plt.ylabel("Dollars")
plt.legend()
plt.grid(True)

plt.show()
```

```python
## 22. scatter plot with trend line
# this connects plotting to modeling.
import matplotlib.pyplot as plt
import numpy as np

hours_studied = np.array([1, 2, 3, 4, 5, 6, 7, 8])
exam_scores = np.array([52, 58, 63, 70, 74, 81, 85, 91])

slope, intercept = np.polyfit(hours_studied, exam_scores, 1)
trend_line = slope * hours_studied + intercept

plt.figure(figsize=(8, 5))

plt.scatter(hours_studied, exam_scores, label="Students")
plt.plot(hours_studied, trend_line, label="Trend line")

plt.title("Study Time vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.legend()
plt.grid(True)

plt.show()
```

```python
## 23. grouped bar chart
# grouped bars compare categories across groups.

import matplotlib.pyplot as plt
import numpy as np

categories = ["Food", "Rent", "Transport", "Entertainment"]
january = [400, 1200, 250, 200]
february = [450, 1200, 300, 180]

x = np.arange(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(x - width / 2, january, width, label="January")
ax.bar(x + width / 2, february, width, label="February")

ax.set_title("Monthly Spending by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Amount")
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()

plt.tight_layout()
plt.show()
```

```python
## 24. dual-axis chart
# dual-axis charts can be useful, 
# but they should be used carefully because they can be misleading.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
revenue = [12000, 15000, 14000, 18000, 21000]
customers = [200, 240, 230, 280, 310]

fig, ax1 = plt.subplots(figsize=(9, 5))

ax1.plot(months, revenue, marker="o")
ax1.set_xlabel("Month")
ax1.set_ylabel("Revenue")

ax2 = ax1.twinx()
ax2.plot(months, customers, marker="s")
ax2.set_ylabel("Customers")

plt.title("Revenue and Customers Over Time")

plt.show()
```

```python
## 25. time series with dates
# dates often require rotated labels and careful spacing.
import matplotlib.pyplot as plt
import pandas as pd

dates = pd.date_range(start="2026-01-01", periods=10, freq="D")
values = [100, 105, 102, 108, 115, 117, 120, 118, 125, 130]

plt.figure(figsize=(10, 5))

plt.plot(dates, values, marker="o")

plt.title("Daily Website Visits")
plt.xlabel("Date")
plt.ylabel("Visits")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
## 26. subplot dashboard with pandas
# This is a good example because it combines:
# subplots, pandas, line charts, bar charts, scatter plots, 
# calculated columns, and dashboard thinking.

import pandas as pd
import matplotlib.pyplot as plt

data = {
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "revenue": [12000, 15000, 14000, 18000, 21000],
    "expenses": [8000, 9000, 9500, 10000, 12000],
    "customers": [200, 240, 230, 280, 310]
}

df = pd.DataFrame(data)
df["profit"] = df["revenue"] - df["expenses"]

fig, axes = plt.subplots(2, 2, figsize=(11, 8))

axes[0, 0].plot(df["month"], df["revenue"], marker="o")
axes[0, 0].set_title("Revenue")

axes[0, 1].plot(df["month"], df["expenses"], marker="o")
axes[0, 1].set_title("Expenses")

axes[1, 0].bar(df["month"], df["profit"])
axes[1, 0].set_title("Profit")

axes[1, 1].scatter(df["customers"], df["revenue"])
axes[1, 1].set_title("Customers vs Revenue")
axes[1, 1].set_xlabel("Customers")
axes[1, 1].set_ylabel("Revenue")

for ax in axes.flat:
    ax.grid(True)

plt.tight_layout()
plt.show()
```
