```python
## 2. CREATE A SMALL DATAFRAME
```

```python
import pandas as pd

# Rows are observations and columns are variables
data = {
    "name": ["Alice", "Bob", "Carlos", "Diana"],
    "major": ["CS", "Math", "Biology", "CS"],
    "exam1": [85, 72, 91, 64],
    "exam2": [88, 75, 94, 70]
}

df = pd.DataFrame(data)

print(df)
```

```python
# view the data "First 5", 5 default

print(df.head())
print(df.head(2))
```

```python
## 3. VIEW THE DATA
```

```python
# view the data "Last 5", 5 default

print(df.tail())
print(df.tail(2))
```

```python
# get shape of the data

print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
## 4. SELECTING COLUMNS
```

```python
names = df["name"]

print(names)
```

```python
grades = df[["name", "exam1", "exam2"]] # select multiple columns

print(grades)
```

```python
## 4. SELECTING A NEW COLUMN
```

```python
df["average"] = (df["exam1"] + df["exam2"]) / 2

# since pandas is built on top of numpy, math happens column by column, automatically

print(df)
```

```python
## 5. FILTER ROWS
```

```python
passed = df[df["average"] >= 70] # students who passed

print(passed)
```

```python
cs_students = df[df["major"] == "CS"] # only CS students

print(cs_students)
```

```python
cs_passed = df[(df["major"] == "CS") & (df["average"] >= 70)] # CS students who passed

## NOTE: In pandas filters, use parentheses around each condition.

print(cs_passed)
```

```python
## 6. SORT DATA
```

```python
sorted_df = df.sort_values("average")

print(sorted_df)
```

```python
sorted_df = df.sort_values("average", ascending=False) # make it descending

print(sorted_df)
```

```python
## 6. GROUP DATA
```

```python
major_averages = df.groupby("major")["average"].mean() # average grade by major

print(major_averages)
```

```python
major_counts = df["major"].value_counts() # count students by major

print(major_counts)
```

```python
### MISSING VALUES ###
```

```python
data = {
    "name": ["Alice", "Bob", "Carlos", "Diana"],
    "exam1": [85, 72, None, 64],
    "exam2": [88, None, 94, 70]
}

df = pd.DataFrame(data)

print(df)
```

```python
# CHECK MISSING VALUES
print(df.isna())
```

```python
# COUNT MISSING VALUES BY COLUMN
print(df.isna().sum())
```

```python
# DROP ROWS WITH MISSING VALUES
clean_df = df.dropna()

print(clean_df)
```

```python
# FILL MISSING VALUES
df["exam1"] = df["exam1"].fillna(df["exam1"].mean())
df["exam2"] = df["exam2"].fillna(df["exam2"].mean())

print(df)
```
